1. Text File (.txt)

```python
# Create a sample text file

text_data = """Name: Rahul
Age: 21
City: Mumbai
Course: Data Science
Name: Priya
Age: 22
City: Pune
Course: Data Science
"""

with open("students.txt", "w") as file:
    file.write(text_data)

print("Text file created successfully.")
```

```python
# Read the text file

with open("students.txt", "r") as file:
    data = file.read()

print(data)
```

```python
# Extract information from the text file

with open("students.txt", "r") as file:
    lines = file.readlines()

names = []
ages = []
cities = []
courses = []

for line in lines:
    line = line.strip()

    if line.startswith("Name:"):
        names.append(line.split(":")[1].strip())

    elif line.startswith("Age:"):
        ages.append(line.split(":")[1].strip())

    elif line.startswith("City:"):
        cities.append(line.split(":")[1].strip())

    elif line.startswith("Course:"):
        courses.append(line.split(":")[1].strip())

print("Names:", names)
print("Ages:", ages)
print("Cities:", cities)
print("Courses:", courses)
```

```python
# Check for missing values

print("Missing Names:", sum(1 for x in names if x == ""))
print("Missing Ages:", sum(1 for x in ages if x == ""))
print("Missing Cities:", sum(1 for x in cities if x == ""))
print("Missing Courses:", sum(1 for x in courses if x == ""))
```

```python
# Convert ages into integers

ages = [int(age) for age in ages]

# Check for age anomalies

anomalies = [age for age in ages if age < 0 or age > 100]

print("Age anomalies:", anomalies)

if len(anomalies) == 0:
    print("No age anomalies found.")
else:
    print("Anomalies found:", anomalies)
```

4. XML File

```python
xml_data = """<?xml version="1.0"?>

<students>

    <student>
        <name>Rahul</name>
        <age>21</age>
        <city>Mumbai</city>
        <marks>85</marks>
    </student>

    <student>
        <name>Priya</name>
        <age>22</age>
        <city>Pune</city>
        <marks>90</marks>
    </student>

    <student>
        <name>Amit</name>
        <age>20</age>
        <city>Nashik</city>
        <marks>78</marks>
    </student>

</students>
"""

with open("students.xml", "w") as file:
    file.write(xml_data)

print("XML file created successfully.")
```

```python
import xml.etree.ElementTree as ET

tree = ET.parse("students.xml")
root = tree.getroot()

print("Root element:", root.tag)
```

```python
names = []
ages = []
cities = []
marks = []

for student in root.findall("student"):

    names.append(student.find("name").text)
    ages.append(int(student.find("age").text))
    cities.append(student.find("city").text)
    marks.append(int(student.find("marks").text))

print("Names:", names)
print("Ages:", ages)
print("Cities:", cities)
print("Marks:", marks)
```

```python
xml_df = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "City": cities,
    "Marks": marks
})

print(xml_df)
```

```python
print("Missing values:")
print(xml_df.isnull().sum())
```

```python
print("Age anomalies:")
print(xml_df[(xml_df["Age"] < 0) | (xml_df["Age"] > 100)])

print("Marks anomalies:")
print(xml_df[(xml_df["Marks"] < 0) | (xml_df["Marks"] > 100)])
```

**2.Write programs for reading and writing binary files.**

```python
import pickle
```

```python
# Sample student data

student = {
    "Name": "Rahul",
    "Age": 21,
    "Course": "Data Science",
    "Marks": 85
}

print("Original Data:")
print(student)
```

```python
# Writing data to a binary file

with open("student.dat", "wb") as file:
    pickle.dump(student, file)

print("Data successfully written to binary file.")
```

```python
# Reading data from binary file

with open("student.dat", "rb") as file:
    data = pickle.load(file)

print("Data read from binary file:")
print(data)
```

**3. Write programs for searching, splitting, and replacing strings based on pattern matching using regular expressions.**

```python
import re
```

Searching using Regular Expression

```python
import re

text = "Contact Rahul at rahul@gmail.com for more information."

pattern = r"\w+@\w+\.\w+"

result = re.search(pattern, text)

if result:
    print("Email found:", result.group())
else:
    print("Email not found.")
```

Splitting using Regular Expression

```python
import re

text = "Rahul, Priya; Amit, Sneha; Rohan"

result = re.split(r"[,;]", text)

print("After splitting:")
print(result)
```

Replacing using Regular Expression

```python
import re

text = "Rahul scored 85 marks and Priya scored 90 marks."

result = re.sub(r"\d+", "XXX", text)

print("Original text:")
print(text)

print("\nAfter replacement:")
print(result)
```

Complete Program

```python
import re

# Sample text
text = "Rahul scored 85 marks. Contact him at rahul@gmail.com; Priya scored 90 marks."

# 1. Searching
pattern = r"\w+@\w+\.\w+"

result = re.search(pattern, text)

if result:
    print("Email found:", result.group())
else:
    print("Email not found.")

# 2. Splitting
split_result = re.split(r"[;]", text)

print("\nAfter splitting:")
print(split_result)

# 3. Replacing
replace_result = re.sub(r"\d+", "XXX", text)

print("\nAfter replacing numbers:")
print(replace_result)
```

Final Complete Program

```python
import sqlite3

# Connect to database
connection = sqlite3.connect("college.db")
cursor = connection.cursor()

# CREATE TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    course TEXT,
    marks INTEGER
)
""")

connection.commit()

print("Table created successfully.")

# INSERT RECORDS
students = [
    (1, "Rahul", 21, "Data Science", 85),
    (2, "Priya", 22, "Computer Science", 90),
    (3, "Amit", 20, "Data Science", 78),
    (4, "Sneha", 21, "Artificial Intelligence", 88)
]

cursor.executemany("""
INSERT OR IGNORE INTO students
(id, name, age, course, marks)
VALUES (?, ?, ?, ?, ?)
""", students)

connection.commit()

print("Records inserted successfully.")

# CREATE - Insert a new record
cursor.execute("""
INSERT OR IGNORE INTO students
(id, name, age, course, marks)
VALUES (5, 'Rohan', 22, 'Data Science', 82)
""")

connection.commit()

print("\nCREATE operation completed.")

# READ
print("\nREAD operation:")
cursor.execute("SELECT * FROM students")

for record in cursor.fetchall():
    print(record)

# UPDATE
cursor.execute("""
UPDATE students
SET marks = 92
WHERE id = 1
""")

connection.commit()

print("\nUPDATE operation completed.")

# Display updated record
cursor.execute("SELECT * FROM students WHERE id = 1")
print("Updated record:", cursor.fetchone())

# DELETE
cursor.execute("""
DELETE FROM students
WHERE id = 5
""")

connection.commit()

print("\nDELETE operation completed.")

# Display final records
print("\nFinal records:")
cursor.execute("SELECT * FROM students")

for record in cursor.fetchall():
    print(record)

# Close connection
connection.close()

print("\nDatabase connection closed.")
```
