```python
# ============================================================
# Practical No. 8: PySpark DataFrame Operations
# ============================================================

# Import required libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, sum, count


# ------------------------------------------------------------
# Step 1: Create Spark Session
# ------------------------------------------------------------

spark = SparkSession.builder \
    .appName("PySpark Practical 8") \
    .getOrCreate()


# ------------------------------------------------------------
# Step 2: Create Sample Sales CSV Dataset
# ------------------------------------------------------------

sales_data = [
    (1, "Laptop", "Electronics", 50000),
    (2, "Mobile", "Electronics", 20000),
    (3, "Chair", "Furniture", 5000),
    (4, "Table", "Furniture", 8000),
    (5, "Laptop", "Electronics", 50000),
    (6, "Headphones", "Electronics", 3000),
    (7, "Chair", "Furniture", 5000)
]

sales_columns = ["id", "product", "category", "sales"]

sales_df = spark.createDataFrame(
    sales_data,
    sales_columns
)

# Save dataset as CSV
sales_df.write.mode("overwrite").option(
    "header", True
).csv("sales_data.csv")


# ------------------------------------------------------------
# Step 3: Read CSV Dataset
# ------------------------------------------------------------

df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("sales_data.csv")

print("Sales Dataset:")
df.show()


# ------------------------------------------------------------
# Step 4: Display Schema
# ------------------------------------------------------------

print("Dataset Schema:")
df.printSchema()


# ------------------------------------------------------------
# Step 5: Filtering
# Display products with sales greater than 10000
# ------------------------------------------------------------

print("Filtered Data:")
df.filter(df.sales > 10000).show()


# ------------------------------------------------------------
# Step 6: Grouping and Aggregation
# Calculate total sales by category
# ------------------------------------------------------------

print("Total Sales by Category:")

df.groupBy("category") \
    .agg(sum("sales").alias("total_sales")) \
    .show()


# ------------------------------------------------------------
# Step 7: Remove Duplicate Records
# ------------------------------------------------------------

print("Data After Removing Duplicates:")

unique_df = df.dropDuplicates()

unique_df.show()


# ------------------------------------------------------------
# Step 8: Create Second Dataset for Join
# ------------------------------------------------------------

product_data = [
    ("Laptop", "Dell"),
    ("Mobile", "Samsung"),
    ("Chair", "IKEA"),
    ("Table", "IKEA"),
    ("Headphones", "Sony")
]

product_columns = ["product", "brand"]

product_df = spark.createDataFrame(
    product_data,
    product_columns
)

print("Product Details:")
product_df.show()


# ------------------------------------------------------------
# Step 9: Join Two Datasets
# ------------------------------------------------------------

print("Joined Dataset:")

joined_df = unique_df.join(
    product_df,
    on="product",
    how="inner"
)

joined_df.show()


# ------------------------------------------------------------
# Step 10: Calculate Average Sales by Product Category
# ------------------------------------------------------------

print("Average Sales by Product Category:")

df.groupBy("category") \
    .agg(avg("sales").alias("average_sales")) \
    .show()


# ------------------------------------------------------------
# Step 11: Stop Spark Session
# ------------------------------------------------------------

spark.stop()

print("PySpark practical completed successfully!")
```
